```python
ans = input("숫자 입력 : ")
l = len(ans)
hol, zak = [], []

for i in ans:
    if int(i) % 2 == 0:
        zak.append(i)
    else:
        hol.append(i)

hol.sort()
zak.sort(reverse=True)

for i in range(l//2):
    print(hol[i], zak[i], end=' ')
```

# 텍스트 파일 분류는 건너뛸거야!!!!!!!!!!

```python
ans = input("현재 시간 (hh:mm:ss 형식으로 입력)")
h, m, s = map(int, ans.split(':'))

if h >= 18:
    print("퇴근할 시간입니다.")
else:
    print(f"남은시간 {17-h}:{59-m}:{60-s}")
```

```python
num = [rd.randint(0,9) for _ in range(3)]
print(num)

while True:
    ans = int(input("숫자 세자리를 입력해주세요."))
    if ans < 100 or ans >= 1000:
        print("세자리가 아냐요")
        continue

    ans_li = [ans//100, (ans%100)//10, ans%10]
    num_dum = num.copy()
    S, B = 0, 0
    for i in range(3):
        if ans_li[i] == num[i]:
            S += 1
        if ans_li[i] in num_dum:
            B += 1
            num_dum.remove(ans_li[i])

    if S == 3:
        print("정답!!")
        break
    print(f"{S}S {B}B")
```
